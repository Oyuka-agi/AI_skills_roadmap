```python
# 1. 从你刚才建的文件中导入模型类
from skill_model import HardSkillExtractionModel

# 2. 定义提取函数 (直接用上一个回答里给你写的函数)
def extract_job_skills(job_text, extraction_model=None, model_path="hard_skill_model.pkl"):
    if not job_text or not isinstance(job_text, str) or len(job_text.strip()) == 0:
        return []

    if extraction_model is None:
        try:
            extraction_model = HardSkillExtractionModel()
            extraction_model.load_model(model_path)
        except Exception as e:
            print(f"❌ Error loading model: {e}")
            return []

    return extraction_model.predict(job_text)

# =====================================================================
# 3. 实际业务调用演示
# =====================================================================
if __name__ == "__main__":
    import json
    
    # 提前加载好模型 (推荐！这样处理多条数据时不会反复读取硬盘)
    print("Loading model for production use...")
    prod_model = HardSkillExtractionModel()
    prod_model.load_model("hard_skill_model.pkl")
    
    # 模拟前端传来的新 Job Post
    new_job_text = """
About the job:
We are looking for a Machine Learning Engineer to join our core AI team. You will be responsible for designing, developing, and deploying scalable machine learning models that drive our business solutions.

Key Responsibilities:
- Work closely with the product team in an Agile environment to understand business needs.
- Build and optimize data pipelines and architecture.
- Train and fine-tune deep learning models for NLP and Computer Vision tasks.

Basic Qualifications:
- 3+ years of experience in software development and data analysis.
- Advanced proficiency in Python and SQL databases.
- Deep understanding of machine learning frameworks such as TensorFlow or PyTorch.
- Extensive experience deploying models to production using AWS (SageMaker, EC2, S3).
- Familiarity with containerization tools like Docker and Kubernetes.
- Basic understanding of CI/CD pipelines and Git version control.
- Strong communication skills and ability to collaborate effectively.
"""
    
    # 调用函数，传入文本和已经加载好的模型
    print("\n--- Extracting Skills ---")
    results = extract_job_skills(new_job_text, extraction_model=prod_model)
    
    # 打印给下游系统（比如匹配学习资源）的 JSON
    print(json.dumps(results, indent=4))
```
